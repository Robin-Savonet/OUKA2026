```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

plt.rcParams["font.family"]     = "serif"
plt.rcParams["font.serif"]      = ["DejaVu Serif"]
plt.rcParams["text.usetex"] = True
plt.rcParams["xtick.direction"] = "in"
plt.rcParams["ytick.direction"] = "in"
plt.rcParams["xtick.top"]       = True
plt.rcParams["ytick.right"]     = True
plt.rcParams["grid.linestyle"]  = "--"
plt.rcParams["grid.alpha"]      = 0.4
plt.rcParams["xtick.labelsize"] = 14
plt.rcParams["ytick.labelsize"] = 14
METHODS = [
    dict(name="ANOVA",         file="ANOVA_all.txt",         mode="max",
         ylabel="F-statistic",   color="#1F77B4"),
    dict(name="FALC",          file="FALC_all.txt",          mode="min",
         ylabel="Residuals",     color="#D62728"),
    dict(name="Lomb-Scargle",  file="Lomb-Scargle_all.txt",  mode="max",
         ylabel="Power",         color="#2ca02c"),
    dict(name="PDM",           file="PDM_all.txt",           mode="min",
         ylabel=r"$\Theta$",     color="#9467bd"),
    dict(name="Dworetsky",     file="Dworetsky_all.txt",     mode="min",
         ylabel="String length", color="#8c564b"),
    dict(name="Lafler-Kinman", file="Lafler-Kinman_all.txt", mode="min",
         ylabel=r"$\Theta$",     color="#e377c2"),
]

# -- Combined figure
fig, axes = plt.subplots(3, 2, figsize=(12, 10))
axes = axes.flatten()

for ax, m in zip(axes, METHODS):
    data   = np.loadtxt(f"{m['file']}")
    P      = data[:, 0] * 24.0   # convert days -> hours
    Y      = data[:, 1]
    idx    = np.argmax(Y) if m["mode"] == "max" else np.argmin(Y)
    P_best = P[idx];  Y_best = Y[idx]

    ax.plot(P, Y, color=m["color"], lw=0.8)
    ax.axvline(P_best, color="black", lw=1.2, ls="--", zorder=5)

    yrange = Y.max() - Y.min()
    ytext  = Y_best - 0.08*yrange if m["mode"]=="max" else Y_best + 0.08*yrange
    va     = "top" if m["mode"]=="max" else "bottom"
    ax.text(P_best + 0.1, ytext, f"  $P = {P_best:.3f}$ h",
            color="black", fontsize=9, va=va, ha="left")

    ax.set_xlabel("Period (h)", fontsize=15)
    ax.set_ylabel(m["ylabel"], fontsize=15)
    ax.set_title(m["name"], fontsize=15)
    ax.grid(True)

fig.tight_layout()
fig.savefig("plots_all/periodograms_all.pdf", dpi=200, bbox_inches="tight")
print("Saved periodograms_all.pdf")

# -- Individual figures
for m in METHODS:
    data   = np.loadtxt(f"{m['file']}")
    P      = data[:, 0] * 24.0   # convert days -> hours
    Y      = data[:, 1]
    idx    = np.argmax(Y) if m["mode"] == "max" else np.argmin(Y)
    P_best = P[idx];  Y_best = Y[idx]

    fig2, ax2 = plt.subplots(figsize=(7, 4))
    ax2.plot(P, Y, color=m["color"], lw=0.8)
    ax2.axvline(P_best, color="black", lw=1.2, ls="--", zorder=5)

    yrange = Y.max() - Y.min()
    ytext  = Y_best - 0.08*yrange if m["mode"]=="max" else Y_best + 0.08*yrange
    va     = "top" if m["mode"]=="max" else "bottom"
    ax2.text(P_best + 0.1, ytext, f"  $P = {P_best:.3f}$ h",
             color="black", fontsize=15, va=va, ha="left")

    ax2.set_xlabel("Period (h)", fontsize=15)
    ax2.set_ylabel(m["ylabel"], fontsize=15)
    #ax2.set_title(m["name"], fontsize=13)
    ax2.grid(True)
    fig2.tight_layout()
    fname = m["name"].replace("-","_").replace(" ","_")
    fig2.savefig(f"plots_all/{fname}_FD58_all.pdf", dpi=200, bbox_inches="tight")
    print(f"  P_best = {P_best:.3f} h  ->  saved {fname}.pdf")

plt.close("all")
```
